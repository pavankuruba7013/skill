```python
n=5
for i in range(1, n+1):
    for j in range(1, n+1):
        #print("*",end = " ")
        if i==1 or i==n or j==1 or j==n:
            print("*", end ="")
        else:
            print(" ", end = "")
    print()
```

```python
n=5
for i in range(1, n+1):
    for j in range(n-i):
        print(" ",end = "")
    for k in range(i):
        print("*", end = "")
    print()
```
